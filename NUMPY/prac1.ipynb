### View बनाम Copy — professional-level बहुत ज़रूरी

```python
x = np.array([10, 20, 30, 40])
a = x[1:3]       # basic slicing -> VIEW (shared memory)
a[0] = 999
print(x)         # [10 999 30 40]

b = x[[1, 2]]    # fancy indexing -> COPY
b[0] = 555
print(x[1])      # 999 (नहीं बदला)

c = x[1:3].copy()  # deliberate independent copy
```

### 5. Boolean masks और Fancy Indexing


```python
scores = np.array([0.15, 0.92, 0.62, 0.40])
keep = scores >= 0.50
print(keep)           # [False  True  True False]
print(scores[keep])   # [0.92 0.62]
print(np.where(keep)[0])  # [1 2]

boxes = np.array([[0, 0, 10, 10],
                  [5, 5, 20, 20],
                  [8, 8, 25, 25],
                  [0, 0,  4,  4]], dtype=np.float32)
filtered_boxes = boxes[keep]
print(filtered_boxes.shape)  # (2, 4)
```

### 5. Boolean masks और Fancy Indexing

```python
import numpy as np

# Confidence scores
scores = np.array([0.15, 0.92, 0.62, 0.40])

# 50% confidence threshold
keep = scores >= 0.50

print("Mask:", keep)
print("Scores:", scores[keep])
print("Indices:", np.where(keep)[0])

# Bounding boxes
boxes = np.array([
    [0, 0, 10, 10],
    [5, 5, 20, 20],
    [8, 8, 25, 25],
    [0, 0, 4, 4]
], dtype=np.float32)

# Filter bounding boxes
filtered_boxes = boxes[keep]

print("Filtered boxes:")
print(filtered_boxes)

print("Shape:", filtered_boxes.shape)
```

### 6. axis का अर्थ — सबसे आम confusion

```python
x = np.array([[1, 2, 3],
              [4, 5, 6]])
print(x.sum())         # 21  all elements
print(x.sum(axis=0))   # [5 7 9]   rows को collapse किया
print(x.sum(axis=1))   # [ 6 15]  columns को collapse किया
print(x.mean(axis=0))  # [2.5 3.5 4.5]
print(x.argmax(axis=1))# [2 2]  हर row में largest का index
```
