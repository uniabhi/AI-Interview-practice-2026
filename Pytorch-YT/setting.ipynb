#### Computtation graph - rememeber each steps (pytorch)

```python
x = torch.tensor(2.0)

w = torch.tensor(3.0, requires_grad=True)
y_pred = w * x
print(y_pred)

y_true = torch.tensor(10.0)

loss = ( y_true - y_pred) **2 
print(loss)

loss.backward()
print(w.grad)
print(x.grad) ## because x is not learnable parameters or can't train only W is thats why W participates in b.prop
```

##### new_value = old_value - (learing_rate*gradient)

```python
x = torch.tensor(2.0) ##input fixed or non-learnable parameters

w = torch.tensor(3.0, requires_grad=True)  ##learnable parameters

##corect answer
y_true = torch.tensor(10.0)

y_pred = w * x

lr = 0.1

loss = (y_pred - y_true) ** 2

loss.backward()

print("before update")
print("w: ", w.item())
print("loss :", loss.item())
print("gradient: ", w.grad.item())

with torch.no_grad():  ##do not track
    w -= lr * w.grad

w.grad.zero_()

print("After Update: ")
print("w :", w.item())
```

```python
x = torch.tensor(2.0)
w = torch.tensor(3.0, requires_grad=True)
y_true = torch.tensor(10.0)

loss1 = (w*x - y_true) **2
loss1.backward()
print("After first backard: ", w.grad.item())

w.grad.zero_()

loss2 = (w * x - y_true) **2
loss2.backward()
print("After 2nd backward: ", w.grad.item())
```

## Training Loop

```python
x = torch.tensor([1., 2., 3., 4.])
y_true = torch.tensor([2., 4., 6., 8.])

w = torch.tensor(0.0, requires_grad=True)

lr = 0.1
epochs = 15

for epoch in range(epochs):
    y_pred = w * x

    loss = ((y_pred - y_true)**2).mean()

    loss.backward()

    with torch.no_grad():  ##during b.propagation do not track or create comp_graph
        w-= lr*w.grad

    print(f"Epoch {epoch+1}: w={w.item():.4f}, loss={loss.item():.4f}")
```

```python
import torch.nn as nn
```

```python
class MyModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.Linear= nn.Linear(1,1)

    def forward(self,x):
        return self.Linear(x)

x =  torch.tensor([[1.], [2.], [3.], [4.]])

y = torch.tensor([[2.], [4.], [6.], [8.]])

model = MyModel()

loss_fn = nn.MSELoss()

optimiser = optim.Adam(model.parameters(), lr=0.1)

for epoch in range(10):
    optimiser.zero_grad()
    y_pred = model(x)
    loss = loss_fn(y_pred, y)
    loss.backward()
    optimiser.step()

    print(f"Epoch: {epoch+1}, Loss:{loss.item():.4f}")

torch.save(model.state_dict(), "linear_model.pth")
```
